## Ranked Actions + Reason Codes

The purpose of this playbook is to turn the model predictions into practical actions that an editorial team can use.

The recommendations are ranked based on predicted content decay and expected business value.

| Priority | Action | Reason Code |
|----------|--------|-------------|
| High | Refresh content immediately | DECAY_HIGH |
| High | Update outdated statistics | OUTDATED_CONTENT |
| Medium | Improve title and metadata | LOW_CTR |
| Low | Monitor performance | STABLE |
| None | No action required | HEALTHY |

These recommendations should be treated as decision-support rather than automatic publishing instructions.

```python
import pandas as pd

action_queue = pd.DataFrame({
    "priority": [1, 2, 3, 4, 5],
    "content_id": [1001, 1002, 1003, 1004, 1005],
    "recommended_action": [
        "Refresh content",
        "Update statistics",
        "Improve metadata",
        "Monitor performance",
        "No action"
    ],
    "reason_code": [
        "DECAY_HIGH",
        "OUTDATED_CONTENT",
        "LOW_CTR",
        "STABLE",
        "HEALTHY"
    ]
})

action_queue
```

## Human Review

Every recommendation should be reviewed by a human editor before action is taken.

Editors should verify:

- factual accuracy
- legal information
- medical information
- financial information
- brand consistency

### What should NOT be automated

The system should never automatically:

- publish articles
- delete content
- rewrite legal pages
- update medical advice
- change pricing
- change company messaging

```python
human_review = pd.DataFrame({
    "Human Review Required": [
        "Legal content",
        "Medical content",
        "Financial advice",
        "Brand messaging",
        "AI-generated updates"
    ]
})

no_go = pd.DataFrame({
    "Never Automate": [
        "Publishing articles",
        "Deleting content",
        "Changing prices",
        "Legal decisions",
        "Medical advice"
    ]
})

display(human_review)
display(no_go)
```
